* **How can I do the same operations on many different values?**

```python
odds = [1, 3, 5, 7]
```

```python
print(odds[0])
print(odds[1])
print(odds[2])
print(odds[3])
```

This is a bad approach for three reasons:
1. **Not scalable**.
1. **Difficult to maintain**.
1. **Fragile**.

```python
odds = [1, 3, 5]
```

```python
print(odds[0])
print(odds[1])
print(odds[2])
print(odds[3])
```

Here’s a better approach: a **for loop**

```python
for num in odds:
    print(num)
```

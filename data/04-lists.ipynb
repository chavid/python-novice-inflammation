* **How can I store many values together?**

## Create a python list

```python
odds = [1, 3, 5, 7]
print(odds)
```

## Access list elements

```python
print(odds[0])
```

```python
print(odds[3])
```

```python
print(odds[-1])
```

```python
print(odds[-2])
```

## Modify list elements

```python
odds[-2] = 0
```

```python
print(odds)
```
